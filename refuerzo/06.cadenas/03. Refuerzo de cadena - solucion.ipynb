## Programa 1: de Vector a Matriz 

Dada una cadena de caracteres y un valor `n`, comprueba si la longitud de la cadena es exactamente `n²`.

Si se cumple, muestra los caracteres organizados en **`n` filas y `n` columnas**, simulando una matriz cuadrada. En caso contrario, muestra un mensaje indicando que la cadena no puede representarse como una matriz de tamaño `n x n`.

````python
s = "ABCDEFGHI"
n = 3

if len(s) == n**2:
    i = 0
    while i < len(s):
        print(s[i], end="")
        j = i + 1
        if j % n == 0:
            print()
        i += 1
else:
    print(f"No es una matriz de {n}x{n}")
````

## Programa 2: Sumar precios de productos

Dada una cadena que contiene varios productos con el formato `producto:precio;`, recórrela para localizar cada precio, convertirlo a número y calcular la **suma total de todos los productos**.

Ejemplo de formato:

```python
"Camiseta:25;Pantalón:70;Calcentin:10;"
```

El programa debe mostrar al final la suma de todos los precios encontrados.

````python
cadena = "Camiseta:25;Pantalón:70;Calcentin:10;"

suma = 0
inicio = 0
i = 0
while i < len(cadena):
    if cadena[i] == ";":
        producto_precio = cadena[inicio:i]
        j = 0
        while j < len(producto_precio):
            if producto_precio[j] == ":":
                precio = producto_precio[j+1:j+3]
                precio = int(precio)
                suma += precio
                #print(precio)
            j += 1
        inicio = i + 1
    i += 1
print(suma)
````
