# Tema 5. Mini-práctica: Control de acceso

Escribe un programa que controle el acceso de personas a una discoteca.

El programa debe preguntar repetidamente:
- La edad de la persona.
- Si tiene entrada (`True` o `False`).
- Si viene acompañada (`True` o `False`).

Una persona puede entrar si:
- Tiene entrada, **y**
- Es mayor de edad **o** viene acompañada.

Después de introducir los datos, se guardará el resultado en una variable booleana llamada `puede_entrar`. A continuación, se mostrará su valor.

Ejemplo:
```
Edad: 16
¿Tiene entrada? True
¿Viene acompañada? False
puede_entrar = False
```

El programa debe seguir preguntando por nuevas personas mientras una variable `respuesta` sea `S`.

## Requisitos

En la solución deben aparecer:

- Variables de tipo `bool`.
- Los operadores `and`, `or` y `not`.
- Al menos una comparación (`>=`, `==`, etc.).
- Un `while` cuya condición dependa de un booleano.
- Un `if` que utilice una expresión booleana.

**Extra:** añade una condición para impedir la entrada a una persona aunque cumpla los requisitos anteriores.

### Primera aproximación

````python
respuesta = "S"
while respuesta == "S":
    edad = int(input("Edad: "))
    if edad >= 18:
        puede_entrar = True
    else:
        tiene_entrada_str = input("¿Tiene entrada? (True/False): ") 
        if tiene_entrada_str == "True":
            acompanada_str = input("¿Viene acompañada? (True/False): ")
            if acompanada_str == "True":
                puede_entrar = True
            else:
                puede_entrar = False
        else:
            puede_entrar = False

    print("¿Puede_entrar?: ", puede_entrar)
    respuesta = input("¿Introducir otra persona? (S/N): ")

print("Fin del programa.")
````

## Solución optimizada con operadores booleanos

````python
respuesta = "S"
while respuesta == "S":
    edad = int(input("Edad: "))
    tiene_entrada_str = input("¿Tiene entrada? (True/False): ") 
    acompanada_str = input("¿Viene acompañada? (True/False): ")

    tiene_entrada = tiene_entrada_str == "True"
    # otra opción:
    tiene_entrada = bool(tiene_entrada_str)

    acompanada = acompanada_str == "True"
    # otra opción:
    acompanada = bool(acompanada_str)

    es_mayor = edad >= 18
    puede_entrar = tiene_entrada and (es_mayor or acompanada)
    print("¿Puede_entrar?: ", puede_entrar)
    respuesta = input("¿Introducir otra persona? (S/N): ")

print("Fin del programa.")
````
