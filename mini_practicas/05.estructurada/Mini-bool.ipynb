# Tema 5. Mini-práctica: Control de acceso

Escribe un programa que controle el acceso de personas a una discoteca.

El programa debe preguntar repetidamente:
- La edad de la persona.
- Si tiene entrada (`True` o `False`).
- Si viene acompañada (`True` o `False`).

Una persona puede entrar si:
- Tiene entrada, **y**
- Es mayor de edad **o** viene acompañada.

Después de introducir los datos, se guardará el resultado en una variable booleana llamada `puede_entrar`. A continuación, se mostrará su valor.

Ejemplo:
```
Edad: 16
¿Tiene entrada? True
¿Viene acompañada? False
puede_entrar = False
```

El programa debe seguir preguntando por nuevas personas mientras el usuario introudzca por teclado la letra `S`.

## Requisitos

En la solución deben aparecer:
- Variables de tipo `bool`.
- Los operadores `and` y `or`.
- Al menos una comparación (`>=`, `==`, etc.).
- Un `while` cuya condición dependa de un booleano.
- Un `if` que utilice una expresión booleana.
