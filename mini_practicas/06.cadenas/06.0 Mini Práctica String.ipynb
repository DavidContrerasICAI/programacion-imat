# 1.1 Codifica contraseñas
"Encripta"/Codifica la contraseña que introduzca el usuario según el siguiente criterio: 
- El caracter de la posición i tendrá el valor siguiente: $nuevo\_valor\_ascii_i = valor\_ascii\_caracter_i + 2 * (i + 1)$

### Ejemplos
- "aaaa"-->"cegi"
- "abcd" --> "cfil"

```python
passwd = input("Introduzca contraseña: ")
nueva_passwd = ""
i = 0
while i < len(passwd):
    nueva_passwd += chr(ord(passwd[i]) + 2 * (i+1))
    i += 1

print(nueva_passwd)
```

## 1.2 Descodifica contraseñas
Ahora se recibirá la contraseña codificada y habrá que volver a la original.

### Ejemplos
- "cegi" --> "aaaa"
- "cfil" --> "abcd"

```python
passwd = input("Introduzca contraseña: ")
nueva_passwd = ""
i = 0
while i < len(passwd):
    nueva_passwd += chr(ord(passwd[i]) - 2 * (i+1))
    i += 1

print(nueva_passwd)
```
