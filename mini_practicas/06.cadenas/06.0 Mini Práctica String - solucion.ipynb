## 2. Almacena coordendas en cadena 
Como hemos visto en la prueba corta, podríamos simplificar el alamcenamiento de coordenadas.
Convierte una cadena con la expresión "x,y" a dos enteros x e y.

### Ejemplos
- "3,5" --> x = 3  y = 5
- "12,45" --> x = 12  y = 45

```python
coordenadas_str = "12,45"

coma = False
x_str = ""
y_str = ""

i = 0
while i < len(coordenadas_str):
    if coordenadas_str[i] == ",":
        coma = True
    elif not coma:
        x_str +=  coordenadas_str[i]
    else:
        y_str +=  coordenadas_str[i]
    i += 1
    
x = int(x_str)
y = int(y_str)
print(x, y)
```
