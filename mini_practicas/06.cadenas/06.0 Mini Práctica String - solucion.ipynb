# Mini Práctica 5.0 Cadenas de caracteres

## Codifica contraseñas
"Encripta"/Codifica la contraseña que introduzca el usuario según el siguiente criterio: 
- El caracter de la posición i tendrá el valor siguiente: $nuevo\_valor\_ascii_i = valor\_ascii\_caracter_i + 2 * (i + 1)$

### Ejemplos
- "aaaa"-->"cegi"
- "abcd" --> "cfil"

## Descodifica contraseñas
Ahora se recibirá la contraseña codificada y se devolverá la original.

### Ejemplos
- "cegi" --> "aaaa"
- "cfil" --> "abcd"
