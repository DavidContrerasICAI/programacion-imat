```python
s = "Luis,Ana,Manuel"
resultado = ""
nombre = ""
i = 0
while i < len(s):
    if s[i] == ",":
        resultado += nombre + "-"
        nombre = ""
    else:
        nombre += s[i]
    i += 1

resultado += nombre
print(resultado)
```

```python
s = "LuisAnaManuel"
resultado = ""
nombre = ""
i = 0
while i < len(s):
    if ord(s[i]) >= ord("A") and ord(s[i]) <= ord("Z") and i > 0:
        resultado += nombre + "-" + s[i]
        nombre = ""
    else:
        nombre += s[i]
    i += 1

resultado += nombre
print(resultado)
```

```python
s = "Luis,Ana,Manuel"
lista = []
nombre = ""
i = 0
while i < len(s):
    if s[i] == ",":
        lista.append(nombre)
        nombre = ""
    else:
        nombre += s[i]
    i += 1

lista.append(nombre)
print(lista)
```
