```python
s = "ab2cde6#dfr@"
s = "ab2#cde1@"
claves = "c1d2e3"

subcadena = ""
resultado = ""

i = 0
while i < len(s):
    #print(s[i], ord(s[i]))
    if ord(s[i]) >= ord("a") and ord(s[i]) <= ord("z"):
        subcadena += s[i]
    elif ord(s[i]) >= ord("0") and ord(s[i]) <= ord("9"):
        repeticiones = int(s[i])
        #print(subcadena*repeticiones)
        resultado_subcadena = ""
        j = 0
        while j < repeticiones:
            resultado_subcadena += subcadena
            j += 1
        if s[i+1] == "#":
            print(">>>> INVERSIÓN", resultado_subcadena)
            s_inv = ""
            j = 0
            while j < len(resultado_subcadena):
                s_inv = resultado_subcadena[j] + s_inv 
                j += 1
            resultado_subcadena = s_inv
        elif s[i+1] == "@":
            print(">>>> OFFSET", resultado_subcadena)
            resultado_subcadena_offset = ""
            j = 0
            while j < len(resultado_subcadena):
                k = 0
                while k < len(claves):
                    if resultado_subcadena[j] == claves[k]:
                        offset = int(claves[k+1])
                        new_car = chr(ord(resultado_subcadena[j]) + offset)
                    k += 1
                resultado_subcadena_offset += new_car
                j += 1
            
            resultado_subcadena = resultado_subcadena_offset
            
        resultado += resultado_subcadena
        subcadena = ""
        
    i += 1

print(resultado)
```
