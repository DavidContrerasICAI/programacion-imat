## Conversión de tipos

Funciones que convierten del tipo str al resto de tipos: 
- str a int
- str a float
- str a complex
- ...

Esta acción en los apuntes se conoce como **cast**.
Las funciones que permiten estas transformaciones se llaman excatamente igual que el tipo destino:
- str a int: int("5")

```python
numero_as_str = "5"
print(numero_as_str + "10")
numero_as_int = int(numero_as_str)
print(numero_as_int)
print(numero_as_int + 10)
print("================")
numero_real_as_str = "55.7"
print(numero_real_as_str + "10")
numero_real_as_float = float(numero_real_as_str)
print(numero_real_as_float)
print(numero_real_as_float + 10)
```

## Salida de un bucle a petición del usuario, no por contador

Muchos bucles no tienen por qué trabajar con contadores o repetirse un número conocido de veces.

```python
numero_objetivo = 5
numero_as_str = input("Número: ")
numero_as_int = int(numero_as_str)
while numero_objetivo != numero_as_int:
    print("No acertaste. Sigue jugando...")
    numero_as_str = input("Número: ")
    numero_as_int = int(numero_as_str)

print("Bien! Acertaste. El número era el", numero_objetivo)
```

## Uniendo condiciones

Si quiero que se cumplan varias condiciones a la vez, por ahora se utilizará esta forma sin utilizar AND u OR que no sabemos lo que son.

```python
# Quiero saber si el número del usuario está en un rango cerrado esta vez

numero_minimo = 5
numero_maximo = 7

numero_as_str = input("Número: ")
numero_as_int = int(numero_as_str)

acertado = False
while acertado == False:
    if numero_as_int >= numero_minimo:
        if numero_as_int <= numero_maximo:
            acertado = True
        else:
            acertado = False
    else:
        acertado = False

    if acertado == False:
        print("No está en el rango. Sigue jugando...")
        numero_as_str = input("Número: ")
        numero_as_int = int(numero_as_str)
    else:
        print("Bien! Acertaste. El número", numero_as_int, "está en el rango [", numero_minimo, ",", numero_maximo, "]")
```
