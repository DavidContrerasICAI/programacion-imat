````python
def limpiar_linea(linea:str)->tuple:
    cp = linea[:5]
    ciudad = linea[5:8].capitalize()
    direccion = linea[8:]
    return cp, ciudad, direccion    # Debe ser un bloque de info
````

````python
linea = "28760MADCalle del pez, 45"
cp, ciudad, direccion = limpiar_linea(linea)
print(f"Vive en {ciudad} en el CP {cp} en la dirección {direccion}")
````
