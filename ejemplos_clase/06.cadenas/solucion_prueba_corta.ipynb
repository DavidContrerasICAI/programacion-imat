```python
FILAS_MAX = 20
COLUMNAS_MIN = 20
FILA_COCHE1 = 10
COL_COCHE1 = 7
FILA_COCHE2 = 6
COL_COCHE2 = 3

fila = 0
while fila < FILAS_MAX:
    col = 0
    while col < COLUMNAS_MIN:
        if fila == 0 or fila == FILAS_MAX - 1 or col == 0 or col == COLUMNAS_MIN - 1:
            print("# ", end="")
        elif fila == col:
            print("* ", end="")
        elif fila == FILA_COCHE1 and col == COL_COCHE1 or fila == FILA_COCHE2 and col == COL_COCHE2:
            print("P ", end="")
        else:
            print("_ ", end="")
        col += 1
    print() 
    fila += 1
```
