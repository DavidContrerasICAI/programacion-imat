```python
i = 1
total = 0

while i < 5:
    j = 0

    while j < i:
        if (i + j) % 2 == 0:
            total = total + i
        else:
            total = total + j

        if total > 10:
            j = j + 1

        j = j + 1

    if total % 3 == 0:
        i = i + 2
    else:
        i = i + 1
```
