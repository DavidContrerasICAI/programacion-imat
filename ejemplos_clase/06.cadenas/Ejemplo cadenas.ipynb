```python
nombre = "MARIA"
# MARIA 
# A
# 2
car_objetivo = "A"
coincidencias = 0
i = 0
while i < len(nombre):
    caracter = nombre[i]
    if caracter == car_objetivo:
        coincidencias += 1
    i += 1
print(f"El caracter {car_objetivo} aparece {coincidencias} veces.")
```
