## Optimización de tuplas/listas

```python
alumnos =["Luis", 8, "Ana", 9, "Juana", 3]
nombre_buscar = "Luis"
for alumno in alumnos:
    if alumno == nombre_buscar:
        print("La nota de", nombre_buscar, "es", "no lo sé")
```

```python
alumnos =[("Luis", 8), ("Ana", 9), ("Juana", 3)]
nombre_buscar = "Luis"
for alumno_tupla in alumnos:
    nombre, nota = alumno_tupla    # "Luis", 8 = ("Luis", 8)
    if nombre == nombre_buscar:
        print("La nota de", nombre_buscar, "es", "no lo sé")
```

```python
alumnos =[("Luis", 8), ("Ana", 9), ("Juana", 3)]
nombre_buscar = "Luis"
for alumno_tupla in alumnos:
    nombre, nota = alumno_tupla    # "Luis", 8 = ("Luis", 8)
    if nombre == nombre_buscar:
        print("La nota de", nombre_buscar, "es", nota)
```

```python
alumnos =[("Luis", 8), ("Ana", 9), ("Juana", 3)]
nombre_buscar = "Luis"
for alumno_tupla in alumnos:
    nombre, nota = alumno_tupla    # "Luis", 8 = ("Luis", 8)
    if nombre == nombre_buscar:
        print(f"La nota de {nombre_buscar} es {nota}")
```

```python
alumnos =[("Luis", 8), ("Ana", 9), ("Juana", 3)]
nombre_buscar = "Luis"
for nombre, nota in alumnos:     # "Luis", 8 = ("Luis", 8)
    if nombre == nombre_buscar:
        print(f"La nota de {nombre_buscar} es {nota}")
```

```python
alumnos =[("Luis", 8), ("Ana", 9), ("Juana", 3)]
nombre_buscar = "Luis"
encontrado = False
i = 0
while not encontrado and i < len(alumnos): 
    print("vuelta", i+1)
    alumno_tupla = alumnos[i]
    nombre, nota = alumno_tupla
    if nombre == nombre_buscar:
        print(f"La nota de {nombre_buscar} es
        {nota}")
        encontrado = True
    i += 1
```
