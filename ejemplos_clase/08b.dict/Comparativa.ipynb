## Creación dinámica

### Listas

```python
lista_alumnos = []
lista_alumnos.append("Luis")
lista_alumnos.append("Ana")
lista_alumnos.append("Juana")
lista_alumnos
```

### Tuplas

```python
tupla_alumno = ("Luis", 21)    # No se puede crear de otra forma
```

### Diccionarios

```python
dict_alumnos = {}
dict_alumnos["Luis"] = 21
dict_alumnos["Ana"] = 22
dict_alumnos["Juana"] = 19
dict_alumnos
```

## Modificación

### Listas

```python
lista_alumnos = ["Luis", "Ana", "Juana"]
lista_alumnos[0] = "Luisito Comunica"
lista_alumnos
```

### Tuplas

Ojo porque son **INMUTABLES**

```python
tupla_alumno = ("Luis", 21)
_, edad = ("Luis", 21)
tupla_alumno = ("Luisito Comunica", edad)    # Se crea otra vez con los valores modificados que se desean
tupla_alumno
```

```python
cadena = "Hola"
cadena[0] = "h"
cadena = "h" + cadena[1:]
cadena
```

```python
tupla_alumno = ("Luis", 21)
tupla_alumno[0] = "Luisito Comunica"
```

```python
tupla_alumno = ("Luis", 21, "Calle del Pez", "Madrid")
nombre, edad, calle, ciudad = tupla_alumno 
tupla_alumno = ("Luisito", edad, calle, ciudad)
tupla_alumno
```

```python
tupla_alumno = ("Luis", 21, "Calle del Pez", "Madrid")
nombre, edad, calle, ciudad = tupla_alumno 
nombre = "Luisito"   # Un = siempre denota creación de una nueva variable
tupla_alumno
```

```python
tupla_alumno = ("Luis", 21, "Calle del Pez", "Madrid", [])
nombre, edad, calle, ciudad = tupla_alumno 
nombre = "Luisito"   # Un = siempre denota creación de una nueva variable
tupla_alumno
```

### Diccionarios

```python
dict_alumnos = {"Luis": 21, "Ana": 22, "Juana": 19}
dict_alumnos["Luis"] = 22    # Cumpleaños de Luis
dict_alumnos["Luisito Comunica"] = 25    # Nueva entrada en el diccionario
dict_alumnos
```

Pero si realmente quiero modificar el nombre de `Luis`...

```python
dict_alumnos["Luisito Comunica"] = 22
del dict_alumnos["Luis"]    # Debo borrar la entrada de una clave-valor
dict_alumnos
```

## Borrado

### Listas

```python
lista_alumnos = ["Luis", "Ana", "Juana"]
lista_alumnos.remove("Luis")
lista_alumnos
```

### Tuplas
Ojo porque son **INMUTABLES**

```python
tupla_alumno = ("Luis", 21, "Calle del Pez, 13")
# Si ya no quiero trabajar con la info de la dirección...
nombre, edad, _ = tupla_alumno
tupla_alumno = (nombre, edad)
tupla_alumno
```

### Diccionarios

```python
dict_alumnos = {"Luis": 21, "Ana": 22, "Juana": 19}
del dict_alumnos["Luis"]
dict_alumnos
```
