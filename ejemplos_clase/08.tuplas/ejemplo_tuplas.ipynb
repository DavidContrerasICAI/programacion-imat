```python
alumnos = []
i = 0
while i < 3:
    cadena = input("nombre,edad")
    datos = cadena.split(",")
    nombre = datos[0].strip()
    edad = int(datos[1].strip())
    alumnos.append((nombre, edad))
    i += 1

for persona in alumnos:
    nombre, edad = persona
    if edad >= 18:
        print(nombre, edad)
```

```python
alumnos = [("Luis", 21), ("Ana", 15), ("Luisa", 30)]
for persona in alumnos:     # persona = alumnos[i]
    nombre = persona[0]
    edad = persona[1]
    if edad >= 18:
        print(nombre, edad)
```
