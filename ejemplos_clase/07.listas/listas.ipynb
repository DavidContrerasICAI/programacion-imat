## Hundir la flota
Queremos definir mediante listas una matriz donde definamos con 0s el agua y con 1s la posición de un barco pequeño de dos posiciones de tamaño

```
fila/col ->  0 1 2 3
--------------------
  0          0 1 0 0
  1          0 1 0 0
  2          0 0 0 0
  3          0 0 0 0
```

Partimos de este ejemplo programado en clase. Modifica y refínalo para que se pueda jugar de una forma muy básica:
1. Definimos la matriz posicionando el barco como en el ejemplo de arriba.
2. Pedimos al usuario las coordenadas donde quiere tirar la bomba.
3. Indicamos al usuario si la bomba ha caido en aguao o ha tocado el barco.
4. Seguimos jugando hasta que hunde el barco.

Se creará el agua y el barco ÚNICAMENTE con appends, sin modificar el código de partida:

````python
matriz = []
matriz.append([])
matriz.append([])
matriz.append([])
matriz.append([])
print(matriz)
````

Este código no sirve, eliminar:

````python
matriz.append(1)   # Hemos visto que así no se agregan barcos, ya que no están dentro de las sublistas
print(matriz)
````

````python
matriz = []
matriz.append([0, 1, 0, 0])
matriz.append([0, 1, 0, 0])
matriz.append([0, 0, 0, 0])
matriz.append([0, 0, 0, 0])
print(matriz)
````

````python
i = 0
while i < len(matriz):
    fila = matriz[i]
    j = 0
    while j < len(fila):
        print(f"{fila[j]} ", end="")
        j += 1
    print()
    i += 1
````

## Primera aproximación de Hundir la Flota

Puntos de mejora a practicar por el alumno:
- Pedir número de barcos.
- Automatizar la petición de las coordenadas de esos barcos en listas.

````python
dimension = int(input("Fila y columnas?"))

posicion1 = input("¿donde coloco el barco 1 (fila,col)?")
lista_posicion1 = posicion1.split(",")
fila_usr1:int = int(lista_posicion1[0].strip())
col_usr1:int = int(lista_posicion1[1].strip()) 

posicion2 = input("¿donde coloco el barco 2 (fila,col)?")
lista_posicion2 = posicion2.split(",")
fila_usr2:int = int(lista_posicion2[0].strip())
col_usr2:int = int(lista_posicion2[1].strip()) 

FILAS = dimension
COLUMNAS = dimension
matriz = []

i = 0
while i < FILAS:
    lista = []
    matriz.append(lista)
    j = 0
    while j < COLUMNAS:
        lista.append(0)
        j += 1
    i += 1

matriz[fila_usr1][col_usr1] = 1
matriz[fila_usr2][col_usr2] = 1

i = 0
while i < len(matriz):
    fila = matriz[i]
    j = 0
    while j < len(fila):
        print(f"{fila[j]} ", end="")
        j += 1
    print()
    i += 1
````

## Cadenas vs listas

La lista final tiene 4 x 4 appends

````python
matriz = []
lista = []   # OJO. LA CREACIÓN DE LA LISTA SE HACE FUERA DEL BUCLE

i = 0
while i < FILAS:
    matriz.append(lista)
    j = 0
    while j < COLUMNAS:
        lista.append(0)
        j += 1
    i += 1

lista[8] = 99    # La matriz tiene 4 copias de la misma lista. Si se modifica la lista una vez, se hace para todos los elementos de matriz, ya que es la misma

print(matriz)
````

Los str son inmutables, no se pueden modificar y no se gestionan como las listas en memoria.

````python
matriz = []
cadena = ""

i = 0
while i < FILAS:
    matriz.append(cadena)    # Toma el valor que tuviese cadena en este paso como un copia.
    j = 0
    while j < COLUMNAS:
        cadena += "0"    # Como los str no tienen APPEND, estamos creando nuevas cadenas cada vez
        j += 1
    i += 1

cadena = "99"

print(matriz)
````
