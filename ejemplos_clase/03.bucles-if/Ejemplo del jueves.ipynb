```python
contador = 0
while contador < 3:
    temp_str = input("Por favor, dime la temperatura número " + str(contador + 1))    # input devuleve siempre un str
    temperatura = int(temp_str)
    
    if temperatura > 25:
        print("Hace muuuuucho calor 😅")
    else:
        if temperatura < 15:
            print("No vea quillo que frío hace")
        else:
            print("Temepratura agradable")
    contador = contador + 1
```
