## ESCALABLE

```python
i = 0
while i < 5:
    print("Hola")
    j = 1
    while j <= 5:
        print(j, end="")
        j += 1
    print()
    i = i + 1

i = 0
while i < 5:
    print("Hola")
    j = 1
    while j <= 5:
        print(j, end="")
        j += 1
    print()
    i = i + 1
```

```python
VECES = 3
MIN = 2
MAX = 6

i = 0
while i < VECES:
    print("Hola")
    j = MIN
    while j <= MAX:
        print(j, end="")
        j += 1
    print()
    i = i + 1
```

```python
numero_objetivo = 4
acierto = False
while acierto == False:
    numero = input("Número:")
    if int(numero) == numero_objetivo:
        acierto = True
```

```python
numero_objetivo = 4

numero_str = input("Número:")
numero = int(numero_str)
while numero != numero_objetivo:
    numero_str = input("Número:")
    numero = int(numero_str)
```

## Solo con if

```python
if 1 == 1:
    print("Paso")
if 2 == 2:
    print("Paso")
if 3 == 3:
    print("Paso")
if 4 == 4:
    print("Paso")
if 5 == 5:
    print("Paso")
```

## Con else

```python
if 1 == 1:
    print("Paso")
else:
    if 2 == 2:
        print("Paso")
    else:
        if 3 == 3:
            print("Paso")
        else:
            if 4 == 4:
                print("Paso")
            else:
                if 5 == 5:
                    print("Paso")
```

## Con elif

```python
if 1 == 1:
    print("Paso")
elif 2 == 2:
    print("Paso")
elif 3 == 3:
    print("Paso")
elif 4 == 4:
    print("Paso")
elif 5 == 5:
    print("Paso")
```
