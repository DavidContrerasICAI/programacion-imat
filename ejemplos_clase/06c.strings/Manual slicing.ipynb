## Operaciones a dominar

Se describen las operacion con slicing que debéis dominar.

```python
cadena = "Manuel"                                                                                                  
```

#### Valores por defecto

```python
print(cadena[0:len(cadena):1])
print(cadena[0::1])
print(cadena[::1])
print(cadena[::])
```

#### Steps                                                                                             

```python
print(cadena[::2]) 
print(cadena[::3])                                                                                                
```

#### Con índices directos

```python
cadena = "1234XYZ"
print(cadena[2:6])    # subcadena                                                                                                  
print(cadena[4:])     # del 4º en adelante
print(cadena[:4])     # los 4 primeros caracteres
```

#### Con índices inversos

```python
cadena = "1234XYZ"
print(cadena[:-3])     # Todos menos los dos últimos caracteres
print(cadena[-3:])     # los 3 últimos caracteres                                                                                          
```

```python
cadena[::-1]    # Invertir la cadena                                                                                                 
```

##### No se pedirá nunca

```python
print(cadena[-5:-1])    # subcadena. NO SE UTILIZA NUNCA   
```

## Ejemplos para útiles

- Indicar el número de referencia del producto. Siempre 3 dígitos después de REF.

```python
producto = "REF823Camiseta a rayas"
codigo = producto[3:6]
print(f"El código numérico de {producto} es {codigo}")
```

- Indicar la clave del alumno que firma el informe y su contenido. Siempre los últimos 8 dígitos del final.

```python
mensaje = "... y este el el informe de la práctica.20261234"
SIZE_CLAVE = 8
clave = mensaje[-SIZE_CLAVE:]
informe = mensaje[:-SIZE_CLAVE]
print(f"La clave del alumno es {clave}")
print(f"Contenido del informe: {informe}")
```

- Indicar el tipo de producto y su nombre.

```python
producto = "FUTCamiseta"
SIZE_TIPO = 3
tipo = producto[:SIZE_TIPO]
ropa = producto[SIZE_TIPO:]
print(f"Tipo de deporte: {tipo}")
print(f"Ropa: {ropa}")
```
