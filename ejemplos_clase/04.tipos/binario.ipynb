### Inferencia de tipos en Programación
Al asignar un valor a una variable, el lenguaje de programación le indica al ordenador automáticamente y de forma transparente al programador, la "estructura de bits" que deberá utilizar para almacenar ese tipo de dato/valor.

```python
numero_entero = -45
numero_decimal = 67.5
valor_logico = True
cadena = "Luis"
```

### Tipos de estructuras de tipos de datos

Hablaremos de forma genérica y abstracta, sin particularizar en ningún lenguaje en concreto, por simplificar:

#### Booleanos (bool)
1 bit, ya que se alamcenan los valores True o False, 1 ó 0.
```python
llueve = True    # guardando un 1 ó 0, nos servirá   
ganador = False    # guardando un 1 ó 0, nos servirá 
```

#### Números enteros (int)
Partiendo de números de 8 bits (muy importante, podríamos hablar de 16, 32 ó 64), utilizaríamos el bit de mayor peso para el signo y el resto para el valor. Podríamos alcanzar hasta el valor $2^{n-1} - 1$, $2^{8-1} - 1 = 127$ (todos los bits a 1 para el valor y a cero el bit de signo).

Para representar el valor 5:
```
signo | valor
   _     _ _ _ _ _ _ _
   0     0 0 0 0 1 0 1
```

Para representar el valor 127:
```
signo | valor
   _     _ _ _ _ _ _ _
   0     1 1 1 1 1 1 1
```

si quisiéramos jugar con Python en binario tenemos que saber que Python no expresa ningún bit para el signo. Solo se ponen solo los 1s y 0s.
```python
0b0000101    # Para representar el 5 
0b1111111    # Para representar el 127 
```

Si quisieramos representar en binario un número negativo, nosostros como humanos deberíamos escribir
```python
-0b101     # Para representar el -5 
```
pero internamente (**MUY IMPORTANTE**) el ordenador realiza la operación del $Complemento_2 = Complemento_1 + 1$

```
        signo | valor binario      | valor deciamal
           _     _ _ _ _ _ _ _
Número 5:  0     0 0 0 0 1 0 1           5
C1 de  5:  1     1 1 1 1 0 1 0     (no existe/no interesa)
C2 de  5:  1     1 1 1 1 0 1 1          -5
```

##### Resumen

En Python nosotros escribiremos simplemente 
```python
n = 5
numero = -5
o
numero = -0b101   # Es una forma humana de expresar binario, no cómo lo almacena el ordenador internamente. Para el ordenador el caracter "-" no existe, solo 1s o 0s.
```
pero el ordenador guardará los 1s y 0s del $Complemento_2$ del 5: 11111011.



##### ¿Y cómo pasamos de un número en C2 a decimal?

Teniendo un número binario en $Complemento_2$ de $k$ bits, siendo sus bits $b_{k-1}, b_{k-2}, \dots, b_{1}, b_{0}$:

$$N = -b_{k-1} \cdot 2^{k-1} + \sum_{i=0}^{k-2} b_i \cdot 2^i$$


````python
#                     1        1        1        1        1        0        1       1
numero_menos_5 =  -1*2**7 + 1*2**6 + 1*2**5 + 1*2**4 + 1*2**3 + 0*2**2 + 1*2**1 + 1*2**0
numero_menos_5
````

**Pregunta final**

¿Por qué si escribo en Python el $Complemento_2$ de -5 de arriba `11111011` no me da -5?

````python
0b11111011
````

**Respuesta**

Porque lo que escribo en Python como 0b11111011 es un binario para humanos, donde el signo negativo se expresa como -0b11111011 y no tiene nada que ver con el $Complemento_2$, que es una forma de representación propia de los ordenadores para alamcenar números negativos. El $Complemento_2$ es solo la vía para pasar de lo humano (-5) a lo que entiende la máquina (-5 en binario según C2).

#### Números reales (float)
Se utiliza la notación científica según el estándar IEEE 754 utilizando la siguiente esctrucutra de bits:

- Bit de signo (S): 1 bit.
- Exponente (E): 11 bits.
- Significando o Mantisa: 53 bits (52 bits se almacenan explícitamente).

Este hecho explica por qué se expresan los números en Python o calculadoras de esta forma:

```python
numero = 0.00052
numero = 5.2e-4
```

No tendremos que conocer nada más de estos tipos de datos / estructuras internas del ordenador.

#### Cadenas de caracteres (str)
Un conjunto de bytes (8 bits) que forma una palabra. Cada letra o caracter de la palabra en un byte (8 bits). No tiene sentido hablar de números negativos porque la tabla ASCII empieza en 0 y termina en 255: $2^8-1=255$.

````python
codigo_ascii_A = 65    # Valor decimal de la A
codigo_ascii_n = 110    # Valor decimal de la n
codigo_ascii_a = 97    # Valor decimal de la a (minúscula, son dos caracteres distintos)
nombre = chr(codigo_ascii_A) + chr(codigo_ascii_n) + chr(codigo_ascii_a)
print(nombre)
````
