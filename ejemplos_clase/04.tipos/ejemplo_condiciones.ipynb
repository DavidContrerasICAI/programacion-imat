# Condicionales

```python
fruta1 = "Melón"
fruta2 = "Sandía"
usuario = input("Dame una fruta: ")
if usuario == fruta1:
    print("Has acertado. Era un melón")
else:
    if usuario == fruta2:
        print("Has acertado. Era una sandía")
    else:
        print("No has acertado")
```

# Bucles y condicionales
Se cumple una de las dos condiciones.

*NOTA: Programa mejorable en las próximas semanas*

```python
fruta1 = "Melón"
fruta2 = "Sandía"
contador = 0
while contador < 5:
    fruta_usuario = input("Dame una fruta: ")
    if fruta_usuario == fruta1:
        print("Has acertado. Era un melón")
        contador = 5
    else:
        if fruta_usuario == fruta2:
            print("Has acertado. Era una sandía")
            contador = 5
        else:
            print("No has acertado")
    contador = contador + 1
```

#### If anidados
Se cumple las dos condiciones.

*NOTA: Programa mejorable en las próximas semanas*

```python
# fruta1 = "Melón"
fruta2 = "Sandía"
frutas_acertada = 0
contador = 0
while contador < 5:
    fruta_usuario = input("Dame una fruta: ")
    if fruta_usuario == fruta1:
        print("    Has acertado. Era un melón!")
        frutas_acertada = frutas_acertada + 1
    else:
        if fruta_usuario == fruta2:
            print("    Has acertado. Era una sandía!")
            frutas_acertada = frutas_acertada + 1
        else:
            print("    No has acertado")
    if frutas_acertada == 2:
        print("Ganaste!!!")
        contador = 5
    else:
        if frutas_acertada == 1:
            print("        Acierta la segunda fruta...")
    contador = contador + 1
```

# Bucles anidados

```python
contador = 0
while contador < 5:
    contador2 = 0
    while contador2 < 5:
        print("*", end="")
        contador2 = contador2 + 1
    print()
    contador = contador + 1
```

```python
contador = 0
while contador < 5:
    contador2 = contador
    while contador2 < 5:
        print("*", end="")
        contador2 = contador2 + 1
    print()
    contador = contador + 1
```

```python
contador = 0
while contador < 5:
    contador2 = 5 - contador
    while contador2 < 6:
        print("*", end="")
        contador2 = contador2 + 1
    print()
    contador = contador + 1
```
