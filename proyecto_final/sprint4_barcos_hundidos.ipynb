# Sprint 4. Proyecto Final - Hundir la flota
## Barcos hundidos

Detectaremos cuando un barco se ha hundido. Hasta el momento podíamos tener barcos de un tamaño 3 con sus 3 posiciones tocadas (💥), pero no detectábamos si realmente lo habíamos hundido. Ahora el porgrama detectará si todas sus psoiciones están tocadas y cambiaremos el barco completo a hundido (💀).

Emojis a incorporar:
- 💀: hundido

Para la detección de un barco hundido deberemos incoprorar una lógica adicional distinta a la anterior, ya que la representación y búsqueda en la matriz mediante emojis úncamente, no servirá.

Si únicamente tenemos en las celdas de la matriz un 🚢, se trata de una información parcial y no podremos coocer el estado general del barco. 

**Ojo**: la lóigca de identificar agua alrededor del barco no servirá, ya que podemos tener barcos contiguos.

Lógica a implementar:
- Modelo de datos que represente el estado del juego y genere una independencia de la representación visual. Todas las acciones que ocurran sobre el juego se plasmarán en este modelo de datos. Posteriorment, se implementará una funcionalidad que traslade esa capa de datos al interfaz de la aplicación.
