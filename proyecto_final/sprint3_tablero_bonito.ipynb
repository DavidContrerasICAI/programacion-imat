Emojis a incorporar:
- 🚢: barco
- 🌊: agua
- 💥: tocado
- ❌: disparo fallido

La lógica del programa se incrementará mediante los siguientes requisitos:
1. Modificar el interfaz para sustituir los 1s y 0s por los emojis de barco y agua.
2. Poner el marco del tablero. Ahora el usuario no especidficará coordenadas (0,1) sino A2.
3. Cada representarán los disparos fallidos y los tocados. En esta iteración no diferenciaremos los tocados de los hundidos.
