# Sprint 1. Proyecto Final - Hundir la flota
## Tablero estático

Mediante listas crearemos una matriz donde definamos con 0s el agua y con 1s la posición de un barco pequeño de dos posiciones de tamaño.

El objetivo es llegar a una disposición similar a la siguiente:

```
fila/col ->  0 1 2 3
--------------------
  0          0 1 0 0
  1          0 1 0 0
  2          0 0 0 0
  3          0 0 0 0
```

La lógica del programa será la siguiente:
1. Mediante bucles, crear una matriz completamente vacía con 4 filas y 4 columnas que será el tablero donde colocaremos el barco.
2. Posicionar nuestro único barco en las posiciones fijas 0,1 y 1,1, como en el ejemplo.
3. Mediante `inputs`, pedir al usuario que lance un bomba. El usuario introducirá las coordenadas mediante un str separados por un coma, por ejemplo, "0,3".
4. El juego finalizará cuando se haya hundido el barco.
